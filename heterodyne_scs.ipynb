```python
import numpy as np
from math import comb
```

```python
def spin_coherent_state(j, theta, phi):
    """
    Returns |Omega> in the |j,m> basis as a complex numpy array.
    """
    d = int(2*j + 1)
    m_vals = np.arange(-j, j+1)

    state = np.zeros(d, dtype=complex)

    for idx, m in enumerate(m_vals):
        coeff = np.sqrt(comb(int(2*j), int(j+m)))
        state[idx] = (
            coeff
            * (np.cos(theta/2)**(j+m))
            * (np.sin(theta/2)**(j-m))
            * np.exp(-1j * m * phi)
        )

    return state

def build_spin_coherent_povm(j, points):
    """
    points: iterable of (theta, phi, weight)
    Returns list of POVM elements E_k
    """
    d = int(2*j + 1)
    povm = []

    prefactor = (2*j + 1) / (4*np.pi)

    for theta, phi, w in points:
        ket = spin_coherent_state(j, theta, phi)
        projector = np.outer(ket, ket.conj())
        E_k = prefactor * w * projector
        povm.append(E_k)

    return povm
```
