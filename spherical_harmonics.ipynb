```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import sph_harm, factorial
```

```python
theta = np.pi/2
phi = 0
m = 0
j = 4.5
n = np.arange(10)
def expectation_spherical_tensors(n = [], m = 0, j = 4.5, theta = np.pi/2, phi = 0):
    results = []
    for i in n:
        c = np.sqrt(4*np.pi * factorial(int(2*j)) / (factorial(int(2*j - i)) * (2*i+1)))
        if np.abs(m) > i:
            results.append(c*np.abs(sph_harm(i, i, theta, phi)))
        else:
            results.append(c * np.abs(sph_harm(m, i, theta, phi)))
    return np.array(results)
plt.plot(n, np.abs(expectation_spherical_tensors(n, m, j, theta, phi)))
plt.xlabel('Degree n')
plt.ylabel('$|Y_{m}^{n}(\\theta, \\phi)|$')
```
